# RETAIL SALES ANALYSIS

### Data Analysis Project using Python

**Tools Used:** Python, Pandas, NumPy, Matplotlib, Seaborn

**Dataset:** Superstore Sales Dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("My Retail Sales Project Started!")
print("All libraries are ready.")
```

```python
df = pd.read_csv("data/Sample - Superstore.csv", encoding="latin-1")
df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nColumn Name:")
print(df.columns)
```

```python
df.isnull().sum() 
```

```python
df.info()
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Ship Date"] = pd.to_datetime(df["Ship Date"])

df[["Order Date", "Ship Date"]].dtypes
```

```python
print("Total Sales:" , df["Sales"].sum())
print("Total Profit:" , df["Profit"].sum())
```

```python
category_sales = df.groupby("Category")["Sales"].sum()

print(category_sales)
```

```python
category_sales.plot(kind="bar",color="skyblue")

plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
region_sales = df.groupby("Region")["Sales"].sum()

print(region_sales)
```

```python
region_sales.plot(kind="bar",color="orange")

plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show
```

```python
category_profit = df.groupby("Category")["Profit"].sum()

print(category_profit)
```

```python
subcategory_sales = df.groupby("Sub-Category")["Sales"].sum()

print(subcategory_sales.sort_values(ascending=False))
```

```python
subcategory_sales.plot(kind="bar", figsize=(12,6), color="skyblue")

plt.title("Sales By Sub-Category ")
plt.xlabel("Sub-Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
top_products = df.groupby("Product Name")["Sales"].sum()

top_10_products = top_products.sort_values(ascending=False).head(10)

print(top_10_products)
```

```python
top_10_products.plot(kind="bar", figsize=(14,8), color="orange")

plt.title("Top 10 Products By Sales")
plt.xlabel("Products Name")
plt.ylabel("Total Sales")
plt.xticks(rotation=45, ha="right")
plt.subplots_adjust(bottom=0.35)
plt.show()
```

```python
region_profit = df.groupby("Region")["Profit"].sum()

print(region_profit.sort_values(ascending=False))
```

```python
region_profit.plot(kind="bar", color="lightgreen")

plt.title("Profit By Region")
plt.xlabel("Region")
plt.ylabel("Total Profit")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
monthly_sales = df.groupby(
    df["Order Date"].dt.to_period("M")
)["Sales"].sum()
print(monthly_sales)
```

```python
plt.figure(figsize=(14,6))

plt.plot(monthly_sales.index.to_timestamp(),
        monthly_sales.values,
        marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```
